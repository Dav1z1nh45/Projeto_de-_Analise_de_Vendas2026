# Projeto: Análise de Vendas de uma Loja (versão simplificada)

Primeiro contato com pandas. Vamos criar uma tabela de vendas e responder algumas perguntas simples sobre ela.

## 1. Importando o pandas

```python
import pandas as pd
```

## 2. Criando a tabela de vendas

```python
dados = {
    "produto": ["Caneta", "Caderno", "Lápis", "Borracha", "Mochila", "Estojo"],
    "preço": [2.5, 15.0, 1.0, 1.5, 89.9, 25.0],
    "quantidade": [120, 45, 200, 90, 15, 30]
}

df = pd.DataFrame(dados)
df
```

## 3. Primeira olhada na tabela

```python
df.head()
```

```python
df.shape
```

## 4. Criando a coluna de faturamento

Faturamento = preço × quantidade

```python
df["faturamento"] = df["preço"] * df["quantidade"]
df
```

## 5. Filtrando produtos com preço acima de R$ 10

```python

caros = df[df["preço"] >= 10]
caros

```

## 6. Descobrindo o produto mais vendido (em quantidade)

```python
df.sort_values("quantidade" , ascending = True)
```

```python
mais_vendido = df.sort_values("quantidade" , ascending = False).iloc[0]
print("O produto mais vendido é" , mais_vendido ["produto"])
```

## 7. Gráfico simples

```python
import matplotlib.pyplot as plt
```

```python

df.plot(x="produto" , y="quantidade" , kind="bar", legend = False)
plt.title ("Quantidade vendida por produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade")
plt.xticks(rotation = 0)
plt.tight_layout()
plt.show()
```

## 8. Agora é com vocês! 🎯

**Questão 1:** Qual foi o faturamento total da loja (soma da coluna `faturamento`)?

```python
faturamento_total = df["Faturamento"].sum()
print(f"Faturamento total da loja: R$ {faturamento_total:.2f}")
```

**Questão 2:** Qual foi o produto que MAIS faturou (não é o mesmo que mais vendeu em quantidade)? Dica: use `sort_values` na coluna `faturamento`.

```python
# Ordena e pega a primeira linha (maior faturamento)
mais_lucrativo = df.sort_values("Faturamento", ascending=False).iloc[0]

print("Produto que mais faturou:", mais_lucrativo["produto"])
print("Faturamento:", mais_lucrativo["Faturamento"])
```
